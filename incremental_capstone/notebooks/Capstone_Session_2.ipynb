## 4. Transform Age and Income

According to the dataset description, the `age` variable is stored as age in years divided by 10, while the `income` variable represents family income in units of $10,000.

To convert these variables into their intended units, `age` will be multiplied by 10 and `income` will be multiplied by 10,000.

The original values will first be displayed so that the results can be compared after the transformation.

```python
# Display the first five age and income values before transformation.
# This allows us to compare the original scaled values with the
# converted values after multiplication.
df[["age", "income"]].head()
```

```python
# Convert age from its scaled value to actual age in years.
# For example, an age value of 6.9 becomes 69 years.
df["age"] = df["age"] * 10

# Convert income from units of $10,000 to actual dollar values.
# For example, an income value of 2.8810 becomes $28,810.
df["income"] = df["income"] * 10000
```

```python
# Display the first five age and income values after transformation.
# This verifies that age is now represented in years
# and income is now represented in actual dollar values.
df[["age", "income"]].head()
```

### Age and Income Transformation Results

The `age` and `income` variables were converted from their original scaled units into more interpretable values.

The `age` variable was multiplied by 10 so that it now represents actual age in years. For example, an original value of 6.9 was converted to 69 years.

The `income` variable was multiplied by 10,000 so that it now represents family income in actual U.S. dollars. For example, an original value of 2.8810 was converted to $28,810.

The transformed values were reviewed to confirm that the calculations were applied successfully.

## 5. Basic Statistical Analysis

Basic statistical measures will be calculated for the numerical variables in the dataset. These measures will help summarize the center, spread, and range of the numerical data.

The analysis will include the count, mean, median, standard deviation, minimum, and maximum values for each numerical column.

```python
# Select only the numerical columns from the DataFrame.
# This excludes string variables such as health, gender, and region.
numeric_df = df.select_dtypes(include="number")

# Calculate several basic statistical measures for each numerical column.
basic_stats = numeric_df.agg([
    "count",
    "mean",
    "median",
    "std",
    "min",
    "max"
])

# Display the statistical results.
basic_stats
```

### Basic Statistical Analysis Findings

The numerical variables contain 4,406 observations each, indicating that there are no missing values among the numerical columns.

The average age of individuals in the dataset is approximately 74.02 years, with a median age of 73 years. Ages range from 66 to 109 years, indicating that the dataset represents an older population.

Average family income is approximately $25,271, while the median income is approximately $16,982. Income ranges from -$10,125 to $548,351. The substantially higher mean compared with the median, along with the large maximum value, suggests that the income distribution may be right-skewed. The presence of negative income values should also be considered during further analysis.

Healthcare utilization varies considerably across individuals. Physician office visits (`visits`) average approximately 5.77 visits, with a median of 4 and a maximum of 89. Several other healthcare utilization variables have median values of zero but relatively large maximum values, suggesting that many individuals did not use those services while a smaller number had much higher utilization.

The number of chronic conditions ranges from 0 to 8, with an average of approximately 1.54 conditions. Years of education (`school`) range from 0 to 18, with an average of approximately 10.29 years.

## 7. Statistical Summary Using `describe()`

The Pandas `describe()` method will be used to generate a standard statistical summary of the numerical variables.

The results from `describe()` will then be compared with the basic statistical analysis performed in the previous section.

```python
# Generate descriptive statistics for the numerical columns.
# describe() automatically calculates count, mean, standard deviation,
# minimum, quartiles, and maximum values.
describe_stats = numeric_df.describe()

# Display the descriptive statistics.
describe_stats
```

### Comparison of Basic Statistics and `describe()`

The results produced by `describe()` are consistent with the basic statistical analysis performed previously. The count, mean, standard deviation, minimum, and maximum values are the same in both analyses. The median calculated previously corresponds to the `50%` value reported by `describe()`.

The primary difference is that `describe()` also provides the 25th and 75th percentiles. These quartiles provide additional information about the distribution of the numerical variables.

For age, the 25th percentile is 69 years, the median is 73 years, and the 75th percentile is 78 years. Therefore, the middle 50% of the observed ages fall between approximately 69 and 78 years.

For income, the 25th percentile is approximately $9,121.50, the median is $16,981.50, and the 75th percentile is $31,728.50. The maximum income of $548,351 is substantially greater than the 75th percentile and median, supporting the earlier observation that the income distribution may be right-skewed.

Several healthcare utilization variables, including outpatient, non-physician outpatient, emergency, and hospital visits, have a 75th percentile of zero. This indicates that at least 75% of observations report no utilization for these particular services, while a smaller portion of individuals accounts for the positive values.

Overall, `describe()` confirms the results of the basic statistical analysis while providing additional information about the distribution of the data through quartiles.
