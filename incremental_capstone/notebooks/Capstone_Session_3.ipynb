## 5. Health and Region Pivot Analysis

A pivot table will be created to examine the distribution of individuals across health status and geographic region.

The table will count the number of observations for each combination of `health` and `region`. This provides a clear view of how the health categories are distributed across the four regional categories.

```python
# Create a pivot table showing the number of individuals
# for each combination of health status and region.
health_region_pivot = pd.pivot_table(
    df,
    index="health",
    columns="region",
    values="age",
    aggfunc="count",
    margins=True,
    margins_name="Total"
)

# Improve the labels so the table is easier to read.
health_region_pivot.index.name = "Health Status"
health_region_pivot.columns.name = "Region"

# Display the completed pivot table.
health_region_pivot
```

### Health and Region Pivot Findings

The pivot table shows that average health is the most frequently reported health status in every region. The number of individuals reporting average health is 957 in the Midwest, 694 in the Northeast, 1,237 in the Other region, and 621 in the West.

Excellent health is the least frequently reported category in the Midwest, Northeast, and Other regions, with 90, 57, and 105 individuals respectively. In the West, 91 individuals report excellent health compared with 86 reporting poor health.

The Other region contains the largest raw number of observations across all three health categories. However, these values represent counts rather than percentages. Therefore, the larger counts may reflect differences in the total number of individuals represented by each region and should not by themselves be interpreted as evidence that one region has better or worse health.

### Average Visits by Health Status and Region

The next pivot table examines the average number of healthcare visits for individuals within each health status and region combination.

Unlike the previous pivot table, which counted the number of individuals, this analysis calculates the mean number of visits. This allows healthcare utilization to be compared across health status and geographic region.

```python
# Create a pivot table showing the average number of visits
# for each combination of health status and region.
health_region_visits_pivot = pd.pivot_table(
    df,
    index="health",
    columns="region",
    values="visits",
    aggfunc="mean"
)

# Rename the row and column labels to make
# the output easier for a reader to understand.
health_region_visits_pivot.index.name = "Health Status"
health_region_visits_pivot.columns.name = "Region"

# Display the averages rounded to two decimal places.
health_region_visits_pivot.style.format("{:,.2f}")
```

### Average Visits by Health Status and Region Findings

The pivot table shows a consistent relationship between self-reported health status and average healthcare visits across all four regions.

Individuals reporting poor health have the highest average number of visits in every region. Average visits among individuals reporting poor health range from 8.12 visits in the Midwest to 10.67 visits in the Northeast. The Northeast has the highest average utilization among individuals reporting poor health.

Individuals reporting excellent health have the lowest average number of visits in every region. Average visits for this group range from 3.14 visits in the Other region to 4.02 visits in the Northeast.

Individuals reporting average health fall between the excellent and poor health groups in every region, with average visits ranging from 5.09 in the Other region to 6.50 in the West.

Overall, the results show that poorer self-reported health is associated with greater healthcare utilization in this dataset. Regional differences are also present, although health status appears to show a consistent relationship with the average number of visits across all regions.

## 6. Analysis of Healthcare Visits and Demographic Factors

The dataset contains several variables representing different forms of healthcare utilization. These variables will first be summarized to understand their distributions before examining their relationships with demographic and economic factors such as gender, marital status, education, income, employment, insurance, and Medicaid.

```python
# Create a list containing the healthcare utilization variables.
visit_columns = [
    "visits",
    "nvisits",
    "ovisits",
    "novisits",
    "emergency",
    "hospital"
]

# Calculate descriptive statistics for the visit variables.
visit_summary = df[visit_columns].describe().T

# Keep the statistics that are most useful for this analysis.
visit_summary = visit_summary[
    ["count", "mean", "50%", "std", "min", "max"]
]

# Rename the columns to make the table easier to read.
visit_summary.columns = [
    "Count",
    "Mean",
    "Median",
    "Std. Dev.",
    "Minimum",
    "Maximum"
]

# Give the row labels a descriptive name.
visit_summary.index.name = "Visit Type"

# Display the results with two decimal places.
visit_summary.style.format("{:,.2f}")
```

### Healthcare Visit Summary Findings

All six healthcare utilization variables contain 4,406 observations, confirming that no missing values are present in these measures.

The `visits` variable has the highest average utilization at 5.77 visits, with a median of 4 visits. The maximum value of 89 visits indicates that some individuals have substantially greater healthcare utilization than the typical individual.

The remaining visit variables have medians of zero. `nvisits` has a mean of 1.62, while `ovisits` and `novisits` have means of 0.75 and 0.54 respectively. Emergency visits and hospital stays have the lowest averages at 0.26 and 0.30.

The zero medians indicate that at least half of the individuals recorded no utilization for `nvisits`, `ovisits`, `novisits`, emergency visits, and hospital stays. At the same time, maximum values such as 104 `nvisits`, 141 `ovisits`, and 155 `novisits` show that a smaller number of individuals have very high healthcare utilization.

Overall, the differences between the means, medians, and maximum values suggest that several healthcare utilization variables have right-skewed distributions, where many individuals have low or zero utilization and a smaller group accounts for much higher utilization.

### Healthcare Utilization by Gender

Healthcare utilization will be grouped by gender to compare the average number of visits across the six healthcare utilization measures.

This analysis helps determine whether the observed average utilization differs between male and female individuals in the dataset.

```python
# Group the dataset by gender and calculate the mean
# for each healthcare utilization variable.
visits_by_gender = df.groupby("gender")[visit_columns].mean()

# Improve the table labels for presentation.
visits_by_gender.index.name = "Gender"

# Rename the visit columns to make them easier to read.
visits_by_gender.columns = [
    "Visits",
    "Nvisits",
    "Ovisits",
    "Novisits",
    "Emergency",
    "Hospital"
]

# Display the average values rounded to two decimal places.
visits_by_gender.style.format("{:,.2f}")
```

### Healthcare Utilization by Gender Findings

The analysis shows some differences in average healthcare utilization between female and male individuals.

Females have a higher average number of overall visits, with 6.01 visits compared with 5.42 visits for males. Females also have a higher average number of `nvisits`, at 1.73 compared with 1.45 for males.

Males have higher average `ovisits` and `novisits`, with averages of 0.85 and 0.57 compared with 0.68 and 0.51 for females. Average hospital utilization is also slightly higher among males at 0.31 compared with 0.29 for females.

Emergency utilization is very similar between the two groups, averaging 0.27 for females and 0.25 for males.

Overall, the results indicate modest differences in healthcare utilization by gender. These descriptive results show associations within the dataset but do not establish that gender causes the observed differences.

### School Distribution

The `school` variable represents years of schooling and ranges from 0 to 18 years. Unlike binary categorical variables such as gender or marital status, school is a numerical variable with multiple possible values.

The distribution of school years will first be examined to determine how individuals are distributed across the different education levels before comparing education with healthcare utilization.

```python
# Count the number of individuals at each reported
# number of years of schooling.
school_distribution = df["school"].value_counts().sort_index()

# Convert the Series into a DataFrame so the output
# can be given clearer, presentation-ready labels.
school_distribution = school_distribution.to_frame(
    name="Number of Individuals"
)

# Give the row index a descriptive label.
school_distribution.index.name = "Years of School"

# Display counts using comma separators and no decimal places.
school_distribution.style.format(
    {"Number of Individuals": "{:,.0f}"}
)
```

### School Distribution Findings

The number of years of schooling ranges from 0 to 18 years, with substantial differences in the number of individuals represented at each education level.

Twelve years of schooling is the most common value, representing 1,252 individuals. Eight years of schooling is the second most common value, with 684 individuals. Several other education levels contain considerably fewer observations.

The concentration at 12 years may reflect individuals completing approximately a high-school level of education, although the dataset only identifies the variable as years of schooling and does not directly provide educational credentials.

Because the number of individuals differs substantially across schooling levels, comparisons should consider these differences in group size when interpreting healthcare utilization.

### Average Healthcare Visits by Years of Schooling

The relationship between education and healthcare utilization will be examined by calculating the average number of healthcare visits for each reported number of years of schooling.

The number of individuals in each education group will also be included so that differences in group size can be considered when interpreting the average number of visits.

```python
# Group the dataset by years of schooling.
# Calculate both the number of individuals and
# the average number of healthcare visits in each group.
school_visits = df.groupby("school").agg(
    Individuals=("visits", "count"),
    Average_Visits=("visits", "mean")
)

# Give the row index a descriptive label.
school_visits.index.name = "Years of School"

# Rename the columns for a cleaner presentation.
school_visits.columns = [
    "Number of Individuals",
    "Average Visits"
]

# Display counts as whole numbers and averages
# rounded to two decimal places.
school_visits.style.format({
    "Number of Individuals": "{:,.0f}",
    "Average Visits": "{:,.2f}"
})
```

### School and Healthcare Utilization Findings

Average healthcare visits vary across years of schooling, and the relationship is not perfectly linear.

Individuals with 18 years of schooling have the highest average number of visits at 7.89, followed by individuals with 15 years of schooling at 7.08. Several of the higher schooling levels have average visit counts above the overall dataset average of approximately 5.77 visits.

However, the average number of visits does not consistently increase with every additional year of schooling. For example, individuals with 16 years of schooling average 6.09 visits, which is lower than the averages for 15, 17, and 18 years.

Group sizes also differ substantially. Twelve years of schooling contains the largest group with 1,252 individuals, while some education levels contain relatively few observations. For example, only 13 individuals report one year of schooling.

Overall, the descriptive results suggest some variation in healthcare utilization across education levels, with several higher schooling levels showing greater average utilization. However, the pattern is not strictly linear, and differences in group size should be considered when interpreting the results.

## 7. Age and Gender Distribution

The distribution of individuals by age and gender will be examined by counting the number of male and female individuals represented at each age.

This analysis provides a demographic view of the dataset and allows differences in age distribution between genders to be observed.

```python
# Create the age and gender distribution table.
age_gender_distribution = pd.crosstab(
    df["age"],
    df["gender"],
    margins=True,
    margins_name="Total"
)

# Improve table labels.
age_gender_distribution.index.name = "Age"
age_gender_distribution.columns.name = "Gender"

# Format counts with commas and display ages as whole numbers.
age_gender_distribution.style.format(
    "{:,.0f}"
).format_index(
    lambda value: (
        "Total" if value == "Total"
        else str(int(float(value)))
    ),
    axis=0
)
```

## 9. Income Distribution by Gender

Income will be analyzed separately for female and male individuals using the number of observations, mean income, median income, minimum income, and maximum income.

Both mean and median income are included because the earlier descriptive analysis showed that income is right-skewed. The median can therefore provide useful context when unusually high income values influence the mean.

```python
# Group the dataset by gender and calculate
# descriptive statistics for income.
income_by_gender = df.groupby("gender")["income"].agg(
    Count="count",
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max"
)

# Give the row index a descriptive label.
income_by_gender.index.name = "Gender"

# Rename the columns for presentation.
income_by_gender.columns = [
    "Number of Individuals",
    "Mean Income",
    "Median Income",
    "Minimum Income",
    "Maximum Income"
]

# Display counts as whole numbers and income as currency.
income_by_gender.style.format({
    "Number of Individuals": "{:,.0f}",
    "Mean Income": "${:,.2f}",
    "Median Income": "${:,.2f}",
    "Minimum Income": "${:,.2f}",
    "Maximum Income": "${:,.2f}"
})
```
